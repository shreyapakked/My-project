```python
import random

games_played = 0
games_won = 0
games_lost = 0
high_score = 0
play_again = "yes"

while play_again == "yes":
    games_played = games_played + 1
    
    print("\nChoose difficulty:")
    print("""1. Easy
    maximum=50
    attempt=10""")
    print("""2. Medium
     maximum=300
    attempt=7""")
    print("""3. Hard
     maximum=500
    attempt=5""")

    try:
        ch = int(input("Enter your choice: "))
    except ValueError:
        print("Invalid input! Defaulting to Easy.")
        ch = 1

    if ch == 1:
        maximum = 50
        attempts = 10
    elif ch == 2:
        maximum = 300
        attempts = 7
    elif ch == 3:
        maximum = 500
        attempts = 5
    else:
        print("Invalid choice, defaulting to Easy mode.")
        maximum = 50
        attempts = 10

    a = random.randint(1, maximum)    
    score = 100

    while attempts > 0:
        try:
            guess = int(input("Enter your guess: "))
        except ValueError:
            print("Please enter valid number.")
            continue

        if guess < 1 or guess > maximum:
            print("Please enter a number between 1 and maximum.")
            continue

        if guess > a:
            print("Too high!")
        elif guess < a:
            print("Too low!")
        else:
            print("You are correct! 🎉")
            print("Your score:", score)
            games_won = games_won + 1

            if score > high_score:
                high_score = score
            break

        attempts = attempts - 1
        score = score - 10  
        print("Number of attempts left is:", attempts)
    else:
        print("\nGame over 😔")
        print("Correct number was:", a)
        games_lost = games_lost + 1

    play_again = input("Do you want to play again (yes/no): ").lower()

print("\n========= GAME STATISTICS =========")
print("Games played:", games_played)
print("Games won:", games_won)
print("Games lost:", games_lost)
print("High score:", high_score)

print("Thanks for playing! 🎮🎮")
```
